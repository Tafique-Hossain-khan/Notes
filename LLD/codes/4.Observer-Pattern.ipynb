```python
class Channel:
    def __init__(self):
        self.latest_video = None
    def upload_video(self, title):
        self.latest_video = title

class Subscriber:
    def check_for_update(self,channel):
        if channel.latest_video:
            print(f"New video uploaded: {channel.latest_video}")
        else:
            print("No new videos uploaded yet.")


# Example usage
if __name__ == "__main__":
    channel = Channel()
    subscriber = Subscriber()

    for i in range(3):
        subscriber.check_for_update(channel)

    channel.upload_video("How to Learn Python")
    subscriber.check_for_update(channel)
```

```python
from abc import ABC, abstractmethod

class IObserver(ABC):
    @abstractmethod
    def update(self):
        pass

class IObservable(ABC):
    @abstractmethod
    def add(self, observer: 'IObserver'):
        pass

    @abstractmethod
    def remove(self, observer: 'IObserver'):
        pass

    @abstractmethod
    def notify(self):
        pass
```

```python
class ConcreteObservable(IObservable):
    def __init__(self):
        self._observers = []
        self._state = None

    def add(self, observer: IObserver):
        self._observers.append(observer)

    def remove(self, observer: IObserver):
        self._observers.remove(observer)

    def notify(self):
        for observer in self._observers:
            observer.update()

    def set_state(self, state):
        self._state = state
        self.notify()

    def get_state(self):
        return self._state
    
class ConcreteObserver(IObserver):
    def __init__(self, observable: ConcreteObservable):
        self._observable = observable
        self._observable.add(self)

    def update(self):
        state = self._observable.get_state()
        print(f"Observer notified. New state: {state}")


# Example usage of the Observer pattern
if __name__ == "__main__":
    observable = ConcreteObservable()
    observer1 = ConcreteObserver(observable)
    observer2 = ConcreteObserver(observable)

    observable.set_state("Hello, World!")
    observable.set_state("Hello World!")
```
