````python
class Candidate:
    def __init__(self, name):
        self.name = name

    def reject(self):
        print(f"{self.name} rejected")

    def move_to_next_round(self):
        print(f"{self.name} moved to next round")

    def send_email(self):
        print(f"Email sent to {self.name}")
````

````python
candidate = Candidate("John")

candidate.move_to_next_round()
candidate.send_email()
````

### Problem:
New Requirement

Business says:

Recruiters should be able to create workflows.

Example:

Graduate Hiring Workflow

1. Move Candidate To Next Round
2. Send Email

The recruiter saves the workflow.

Later it gets executed.

First Attempt (Without Command)

Try to model it.

Maybe:
``` python
workflow = [
    candidate.move_to_next_round(),
    candidate.send_email()
]
```
What happens?

Immediately:

John moved to next round
Email sent to John

Oops.

The actions executed.

We wanted to STORE them.

First Pain Point

We want:

Store action

not

Execute action 

````python
workflow = [
    candidate.move_to_next_round(),
    candidate.send_email()
]
````

## Approch 2

````python
workflow = [
    candidate.move_to_next_round,
    candidate.send_email
]
````

````python
for action in workflow:
    action()
````

## Approch 3

````python
# insted of storing the methods in a list, we can create a command class.
class MoveToNextRoundCommand:
    def __init__(self, candidate):
        self.candidate = candidate

    def execute(self):
        self.candidate.move_to_next_round()
````

````python
cmd = MoveToNextRoundCommand(candidate)
````

````python
workflow = [cmd]
````

## If the requirement increased to add more options

````python
from abc import ABC, abstractmethod
class Command(ABC):
    @abstractmethod
    def execute(self):
        pass

class MoveToNextRoundCommand(Command):
    def __init__(self, candidate):
        self.candidate = candidate

    def execute(self):
        self.candidate.move_to_next_round()

class SendEmailCommand(Command):
    def __init__(self, candidate):
        self.candidate = candidate

    def execute(self):
        self.candidate.send_email()

class RejectCommand(Command):
    def __init__(self, candidate):
        self.candidate = candidate

    def execute(self):
        self.candidate.reject()
````

````python
workflow = [
    MoveToNextRoundCommand(candidate),
    SendEmailCommand(candidate),
    RejectCommand(candidate)
]
````
