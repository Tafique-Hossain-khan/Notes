```python
from abc import ABC, abstractmethod

class FileSystemItem(ABC):
    @abstractmethod
    def ls(self, indent=0):
        pass

    @abstractmethod
    def open_all(self, indent=0):
        pass

    @abstractmethod
    def get_size(self):
        pass

    @abstractmethod
    def get_name(self):
        pass

    @abstractmethod
    def cd(self, target_name):
        pass

    @abstractmethod
    def is_folder(self):
        pass
```

### File class implementation

```python
class File(FileSystemItem):
    def __init__(self, name, size):
        self.name = name
        self.size = size

    def ls(self, indent=0):
        print(" " * indent + self.name)

    def open_all(self, indent=0):
        print(" " * indent + self.name)

    def get_size(self):
        return self.size

    def get_name(self):
        return self.name

    def cd(self, target_name):
        return None

    def is_folder(self):
        return False
```

```python
f = File("file1.txt", 1)
f.ls()  # Output: file1.txt
```

```python
f.open_all()
```

```python
f.get_name()  # Output: file1.txt
```

```python
f.get_size()  # Output: 1
```

```python
f.cd("any_folder")  # Output: None
```

```python
class Folder(FileSystemItem):
    def __init__(self, name):
        self.name = name
        self.children = []

    def add(self, item):
        self.children.append(item)

    def ls(self, indent=0):
        for child in self.children:
            prefix = "+ " if child.is_folder() else "  "
            print(" " * indent + prefix + child.get_name())
    
    def open_all(self, indent=0):
        print(" " * indent + self.name + "/")
        for child in self.children:
            child.open_all(indent + 2)

    def get_size(self):
        total_size = 0
        for child in self.children:
            total_size += child.get_size()
        return total_size

    def get_name(self):
        return self.name

    def cd(self, target_name):
        if self.name == target_name:
            return self
        for child in self.children:
            if child.is_folder():
                result = child.cd(target_name)
                if result is not None:
                    return result
        return None

    def is_folder(self):
        return True
```

```python
root = Folder("root")
root.add(File("file1.txt", 1))
root.add(File("file2.txt", 1))

docs = Folder("docs")
docs.add(File("resume.pdf", 1))
docs.add(File("notes.txt", 1))

images = Folder("images")
images.add(File("photo.jpg", 1))

root.add(docs)
root.add(images)
```

```python
def main():
    root = Folder("root")
    root.add(File("file1.txt", 1))
    root.add(File("file2.txt", 1))

    docs = Folder("docs")
    docs.add(File("resume.pdf", 1))
    docs.add(File("notes.txt", 1))

    images = Folder("images")
    images.add(File("photo.jpg", 1))

    root.add(docs)
    root.add(images)

    print("--- open_all ---")
    root.open_all()

    print("--- ls (root) ---")
    root.ls()

    print("--- ls (docs) ---")
    docs.ls()

    print("--- cd docs, then ls ---")
    cw = root.cd("docs")
    if cw is not None:
        cw.ls()
    else:
        print("Could not cd into docs")

    print("--- get_size (root) ---")
    print(root.get_size())

if __name__ == "__main__":
    main()
```
