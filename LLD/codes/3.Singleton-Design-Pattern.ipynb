```python
class Singleton:
    def __init__(self):
        print("Singleton constructor called, new object created")
```

```python
s1 = Singleton()
s2 = Singleton()

print(s1 is s2)
```

# Python does not have true private constructors like C++ or Java.

# The actual object creation happens in __new__(), not __init__().

# __new__() is responsible for:
# 1. Allocating memory for the object.
# 2. Creating the instance.
# 3. Returning the instance.

# __init__() runs AFTER the object has already been created.
# Its job is only to initialize the instance.

# Therefore, the Python equivalent mechanism for controlling
# object creation is overriding __new__().

# By overriding __new__(), we can decide:
# - Create and return a new object.
# - Return an already existing object.
# - Prevent object creation entirely.

# This is the mechanism used by the Singleton pattern in Python.

# Example (concept only):

class Example:
    def __new__(cls):
        print("Object creation happens here")
        return super().__new__(cls)

    def __init__(self):
        print("Object initialization happens here")

# Since __new__() controls instance creation,
# it can be overridden to return an existing object
# instead of creating a new one.
#
# Therefore:
# Yes, by overriding __new__(), you can control whether
# a new object is created or an existing one is returned.

```python
import threading

class Tower:
    _instance = None
    _lock = threading.Lock()

    def __new__(cls):
        if cls._instance is None:
            with cls._lock:
                if cls._instance is None:
                    cls._instance = super().__new__(cls)
        return cls._instance
    
    def return_instance(self):
        return self._instance
```

```python
t1 = Tower()
t2 = Tower()

print(t1 is t2)
```

```python
t1.return_instance()
```

```python
t2.return_instance()
```

```python
import psycopg2.pool
import threading


class DatabasePool:
    _instance = None
    _lock = threading.Lock()

    def __new__(cls):
        if cls._instance is None:
            with cls._lock:
                if cls._instance is None:
                    cls._instance = super().__new__(cls)

                    cls._instance.pool = (
                        psycopg2.pool.SimpleConnectionPool(
                            minconn=1,
                            maxconn=10,
                            host="localhost",
                            database="shop",
                            user="admin",
                            password="secret"
                        )
                    )

        return cls._instance
```
