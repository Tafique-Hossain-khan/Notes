```python
from abc import ABC, abstractmethod


class MoneyHandler(ABC):
    """Abstract base class for all denomination handlers."""

    def __init__(self):
        self._next_handler: 'MoneyHandler | None' = None

    def set_next_handler(self, handler: 'MoneyHandler') -> None:
        """Link the next handler in the chain."""
        self._next_handler = handler

    @abstractmethod
    def dispense(self, amount: int) -> None:
        """Try to dispense the requested amount. Pass remainder to next handler."""
        pass


class Handler1000(MoneyHandler):
    """Handles ₹1000 notes."""

    def __init__(self, num_notes: int):
        super().__init__()
        self._num_notes = num_notes

    def dispense(self, amount: int) -> None:
        notes_needed = amount // 1000

        if notes_needed > self._num_notes:
            notes_needed = self._num_notes
            self._num_notes = 0
        else:
            self._num_notes -= notes_needed

        if notes_needed > 0:
            print(f"Dispensing {notes_needed} x ₹1000 note(s)")

        remaining = amount - (notes_needed * 1000)

        if remaining > 0:
            if self._next_handler is not None:
                self._next_handler.dispense(remaining)
            else:
                print(f"Cannot fulfil remaining ₹{remaining} — insufficient funds in ATM")


class Handler500(MoneyHandler):
    """Handles ₹500 notes."""

    def __init__(self, num_notes: int):
        super().__init__()
        self._num_notes = num_notes

    def dispense(self, amount: int) -> None:
        notes_needed = amount // 500

        if notes_needed > self._num_notes:
            notes_needed = self._num_notes
            self._num_notes = 0
        else:
            self._num_notes -= notes_needed

        if notes_needed > 0:
            print(f"Dispensing {notes_needed} x ₹500 note(s)")

        remaining = amount - (notes_needed * 500)

        if remaining > 0:
            if self._next_handler is not None:
                self._next_handler.dispense(remaining)
            else:
                print(f"Cannot fulfil remaining ₹{remaining} — insufficient funds in ATM")


class Handler200(MoneyHandler):
    """Handles ₹200 notes."""

    def __init__(self, num_notes: int):
        super().__init__()
        self._num_notes = num_notes

    def dispense(self, amount: int) -> None:
        notes_needed = amount // 200

        if notes_needed > self._num_notes:
            notes_needed = self._num_notes
            self._num_notes = 0
        else:
            self._num_notes -= notes_needed

        if notes_needed > 0:
            print(f"Dispensing {notes_needed} x ₹200 note(s)")

        remaining = amount - (notes_needed * 200)

        if remaining > 0:
            if self._next_handler is not None:
                self._next_handler.dispense(remaining)
            else:
                print(f"Cannot fulfil remaining ₹{remaining} — insufficient funds in ATM")


class Handler100(MoneyHandler):
    """Handles ₹100 notes. Acts as the last handler in this chain."""

    def __init__(self, num_notes: int):
        super().__init__()
        self._num_notes = num_notes

    def dispense(self, amount: int) -> None:
        notes_needed = amount // 100

        if notes_needed > self._num_notes:
            notes_needed = self._num_notes
            self._num_notes = 0
        else:
            self._num_notes -= notes_needed

        if notes_needed > 0:
            print(f"Dispensing {notes_needed} x ₹100 note(s)")

        remaining = amount - (notes_needed * 100)

        if remaining > 0:
            # Last handler — no next_handler available
            print(f"Cannot fulfil remaining ₹{remaining} — no smaller denomination available")


# -----------------------------------------------------------------------
# Client / Main
# -----------------------------------------------------------------------
if __name__ == "__main__":
    # Create handlers with available note counts
    h1000 = Handler1000(num_notes=3)   # 3 × ₹1000 = ₹3000 available
    h500  = Handler500(num_notes=5)    # 5 × ₹500  = ₹2500 available
    h200  = Handler200(num_notes=10)   # 10 × ₹200 = ₹2000 available
    h100  = Handler100(num_notes=20)   # 20 × ₹100 = ₹2000 available

    # Build the chain
    h1000.set_next_handler(h500)
    h500.set_next_handler(h200)
    h200.set_next_handler(h100)
    # h100 has no next — it is the last handler

    print("=== Withdraw ₹300 ===")
    h1000.dispense(300)

    print()
    print("=== Withdraw ₹3000 ===")
    h1000.dispense(3000)

    print()
    print("=== Withdraw ₹4000 ===")
    h1000.dispense(4000)

    print()
    print("=== Withdraw ₹150 (no ₹50 handler exists) ===")
    h1000.dispense(150)
```
