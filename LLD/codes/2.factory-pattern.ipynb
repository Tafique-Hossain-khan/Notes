````python
from abc import ABC, abstractmethod
````

#### strategy for ai services

````python
class LLMProvider(ABC):
    @abstractmethod
    def generate_response(self, prompt: str) -> str:
        pass

class OpenAIService(LLMProvider):
    def generate_response(self, prompt: str) -> str:
        # Simulate OpenAI API call
        return f"OpenAI response to '{prompt}'"
class GoogleAIService(LLMProvider):
    def generate_response(self, prompt: str) -> str:
        # Simulate Google AI API call
        return f"Google AI response to '{prompt}'"
class ClaudeAIService(LLMProvider):
    def generate_response(self, prompt: str) -> str:
        # Simulate Claude AI API call
        return f"Claude AI response to '{prompt}'"
    
class AIservice:
    def __init__(self, strategy: LLMProvider):
        self._strategy = strategy

    def generate_response(self, prompt: str) -> str:
        return self._strategy.generate_response(prompt)
````

### Factory for routing

````python
class AIFactory:
    @staticmethod
    def get_ai_service(service_type: str) -> LLMProvider:
        services = {
            "openai": OpenAIService(),
            "google": GoogleAIService(),
            "claude": ClaudeAIService()
        }
        service_class = services.get(service_type)
        if not service_class:
            raise ValueError(f"Unknown AI service type: {service_type}")
        return service_class
````

````python
service:LLMProvider = AIFactory.get_ai_service("openai")

chat = AIservice(service)
response = chat.generate_response("What is the capital of France?")
response
````
