```python
from dataclasses import dataclass
@dataclass
class Candidate:
    name: str
    email: str
```

```python
class EmailService:
    def send_welcome_email(self, candidate:Candidate):
        print(f"Sending welcome email to {candidate.name} at {candidate.email}")
class EmployeeIdGenerator:
    def generate_employee_id(self, candidate:Candidate):
        print(f"Generating employee ID for {candidate.name}")

class SalaryAccountService:
    def create_salary_account(self, candidate:Candidate):
        print(f"Creating salary account for {candidate.name}")
class CompanyEmailService:
    def create_company_email(self, candidate:Candidate):
        print(f"Creating company email for {candidate.name}")

class BackgroundVerificationService:
    def verify(self, candidate:Candidate):
        print(f"Verifying background for {candidate.name}")

class SlackAccountService:
    def create(self, candidate:Candidate):
        print(f"Creating slack account for {candidate.name}")

class InsuranceService:
    def enroll(self, candidate:Candidate):
        print(f"Enrolling {candidate.name} in insurance")



```

```python
class HRFacade:
    def __init__(
        self,
        email_service: EmailService,
        employee_id_generator: EmployeeIdGenerator,
        salary_account_service: SalaryAccountService,
        company_email_service: CompanyEmailService,
        background_verification_service: BackgroundVerificationService,
        slack_account_service: SlackAccountService, 
        insurance_service: InsuranceService
    ):
        self.email_service = email_service
        self.employee_id_generator = employee_id_generator
        self.salary_account_service = salary_account_service
        self.company_email_service = company_email_service
        self.background_verification_service = background_verification_service
        self.slack_account_service = slack_account_service
        self.insurance_service = insurance_service

    def onboard_candidate(self, candidate: Candidate):
        self.employee_id_generator.generate_employee_id(candidate)
        self.salary_account_service.create_salary_account(candidate)
        self.company_email_service.create_company_email(candidate)
        self.email_service.send_welcome_email(candidate)
        self.background_verification_service.verify(candidate)
        self.slack_account_service.create(candidate)
        self.insurance_service.enroll(candidate)
```

```python
candidate = Candidate("John", "john@example.com")

onboarding_facade = HRFacade(
    EmailService(),
    EmployeeIdGenerator(),
    SalaryAccountService(),
    CompanyEmailService(),
    BackgroundVerificationService(),
    SlackAccountService(),
    InsuranceService()
)
onboarding_facade.onboard_candidate(candidate)
```
