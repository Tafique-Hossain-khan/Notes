# Problem Statement

You are building a recruitment platform.

One feature is:

```text id="w9uh3s"
Send interview invitations to candidates.
```

Your application was designed around this interface:

```python id="kjlwmc"
class NotificationService:
    def send(self, to, subject, message):
        pass
```

Application code:

```python id="fxn44v"
notification_service.send(
    "john@example.com",
    "Interview Invitation",
    "Please join tomorrow at 10 AM"
)
```

Everything is working.

````python
class NotificationService:
    def send(self, to, subject, message):
        pass

class EmailNotificationService(NotificationService):
    def send(self, to, subject, message):
        print(f"Sending email to {to} with subject '{subject}' and message '{message}'")
````

````python
notification_service = EmailNotificationService()
````

````python
notification_service.send(
    "john@example.com",
    "Interview Invitation",
    "Please join tomorrow at 10 AM"
)
````

# Bonus Challenge (After You Solve It)

Tomorrow another vendor arrives.

```python id="r5whx0"
class SuperMailer:

    def dispatch(
        self,
        email_address,
        email_subject,
        email_content
    ):
        pass
```

Can your application still use:

```python id="w12d6s"
notification_service.send(...)
```

without changing any existing application code?

If you can solve that cleanly, you'll have discovered the Adapter pattern naturally rather than memorizing its UML.


````python
class NotificationAdapter(NotificationService):
    def __init__(self, third_party_service):
        self.third_party_service = third_party_service

    def send(self, to, subject, message):
        self.third_party_service.send_email(to, subject, message)
````

````python
notify = NotificationAdapter(ThirdPartyEmailProvider())

notify.send(
    "john@example.com",
    "Interview Invitation",
    "Please join tomorrow at 10 AM"
)
````

## New vendor was introduced

````python
class Newvendor:
    def dispatch(self, email_address, title, content):
        print("Email Sent from New Vendor")
````

````python
class EmailNotificationAdapter(NotificationService):
    def __init__(self, vendor):
        self.vendor = vendor
    
    def send(self, to, subject, message):
        self.vendor.dispatch(to, subject, message)
````

````python
email = EmailNotificationAdapter(Newvendor())
email.send("john@example.com", "Interview Invitation", "Please join tomorrow at 10 AM")
````
