## **Option A — Notification System (Easy)**

A `NotificationService` class that sends messages using a notification strategy.

Strategies:

- `EmailNotification`
- `SMSNotification`
- `SlackNotification`

The service holds a notification strategy and exposes a:

```python
send(message: str) -> None
```
method.

Support swapping the notification strategy after creation.

### Example

```python
service = NotificationService(EmailNotification())
service.send("Order shipped")

service.set_strategy(SMSNotification())
service.send("Order shipped")
```


````python
from abc import ABC, abstractmethod
````

````python
class NotificationStrategy(ABC): #use the name that reflects the pattern
    @abstractmethod
    def send_notification(self, message: str):
        pass

class EmailNotificationService(NotificationStrategy):
    def send_notification(self, message: str):
        print(f"Sending email notification: {message}")

class SMSNotificationService(NotificationStrategy):
    def send_notification(self, message: str):
        print(f"Sending SMS notification: {message}")

class PushNotificationService(NotificationStrategy):
    def send_notification(self, message: str):
        print(f"Sending push notification: {message}")

class NotificationService:
    def __init__(self, service: NotificationStrategy):
        self._service = service

    def set_service(self, service: NotificationStrategy):
        self._service = service

    def send(self, message: str):
        self._service.send_notification(message)
````

````python
ns = NotificationService(service=EmailNotificationService())
ns.send("Hello via Email!")
````

````python
ns.set_service(SMSNotificationService())
ns.send("Hello via SMS!")
````

````python
class AIProviderStrategy(ABC):
    @abstractmethod
    def generate_response(self, prompt: str) -> str:
        pass

class OpenAIProviderStrategy(AIProviderStrategy):
    def generate_response(self, prompt: str) -> str:
        return f"OpenAI response to: {prompt}"
class GoogleAIProviderStrategy(AIProviderStrategy):
    def generate_response(self, prompt: str) -> str:
        return f"Google AI response to: {prompt}"   


class AIChatbot:
    def __init__(self, provider: AIProviderStrategy):
        self._provider = provider

    def set_provider(self, provider: AIProviderStrategy):
        self._provider = provider

    def get_response(self, prompt: str) -> str:
        return self._provider.generate_response(prompt)
````

````python
chatbot = AIChatbot(provider=OpenAIProviderStrategy())
print(chatbot.get_response("What is the capital of France?"))
````
